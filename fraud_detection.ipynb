```python
#Fraud rate by transaction type
fraud_rate=(df.groupby("type")["isFraud"].mean().mul(100).round(2).sort_values(ascending=False))
fraud_rate
```

```python
#Fraud amount by transaction type
fraud_amount=df[df["isFraud"]==1].groupby("type")["amount"].sum().sort_values(ascending=False)
fraud_amount
```

```python
#Fraud Rate by transaction amount
bins=[0,10000,50000,100000,500000,float("inf")]
labels=["0-10k","10k-50k","50k-100k","100k-500k","500k+"]
df["amount_range"]=pd.cut(df["amount"],bins=bins,labels=labels,include_lowest=True)
fraud_by_amount=(df.groupby("amount_range",observed=True)["isFraud"].mean().mul(100).round(2))
fraud_by_amount
```

```python
#fraud trend over time
fraud_by_step=(df.groupby("step")["isFraud"].sum().sort_values(ascending=False))
print(fraud_by_step.head(10))
```

```python
step_212=df[(df["step"]==212)&(df["isFraud"]==1)]
step_212
```

```python
#fraud rate by time period
fraud_rate_by_step=(df.groupby("step")["isFraud"].mean().mul(100).round(2).sort_values(ascending=False))
fraud_rate_by_step
```

```python
time_analysis=(df.groupby("step").agg(total_transactions=("isFraud","count"),fraud_transactions=("isFraud","sum")))
time_analysis
```

```python
time_analysis["fraud_rate"]=(time_analysis["fraud_transactions"]/time_analysis["total_transactions"]*100).round(2)
time_analysis
```

```python
time_analysis=time_analysis[time_analysis["total_transactions"]>=100].sort_values("fraud_rate",ascending=False)
time_analysis
```

```python
#identify high risk transaction types and amounts together
fraud_type_amount=((df.groupby(["type","amount_range"],observed=True))["isFraud"].mean().mul(100).round(2).sort_values(ascending=False))
fraud_type_amount.head(10)
```

```python
fraud_amount_by_account=(df[df["isFraud"]==1].groupby("nameOrig")["amount"].sum().sort_values(ascending=False))
fraud_amount_by_account.head(10)                         
```

```python
fraud_account_details=(df[df["isFraud"]==1].groupby("nameOrig").agg(fraud_transactions=("isFraud","sum"),total_fraud_amount=("amount","sum")))
fraud_account_details.sort_values("total_fraud_amount",ascending=False).head(10)
```

```python
fraud_receiver=(df[df["isFraud"]==1].groupby("nameDest").agg(total_fraud_amount=("amount","sum"),fraud_transactions=("isFraud","sum")))
fraud_receiver.sort_values("total_fraud_amount",ascending=False).head(10)
```

```python
fraud_receivers=df[df["isFraud"]==1]["nameDest"]
print("Total_Fradualant_Transactions:",len(fraud_receivers))
print("Unique_receiver_accounts:",fraud_receivers.nunique())
```

```python
#check suspicious balance changes
df["expected_new_balance"]=(df["oldbalanceOrg"]-df["amount"])
df["balance_difference"]=(df["newbalanceOrig"]-df["expected_new_balance"])
print(df[["oldbalanceOrg","amount","newbalanceOrig","expected_new_balance","balance_difference"]].head(10))
```

```python
largest_balance_difference=(df[["oldbalanceOrg","amount","newbalanceOrig","expected_new_balance","balance_difference"]].sort_values("balance_difference",ascending=False))
print(largest_balance_difference.head(20))                           
```

```python
df["balance_mismatch"]=(df["balance_difference"].abs()>1)
df["balance_mismatch"]
```

```python
balance_fraud_rate=(df.groupby("balance_mismatch")["isFraud"].mean().mul(100).round(2))
balance_fraud_rate
```

```python
flagged_analysis=pd.crosstab(df["isFlaggedFraud"],df["isFraud"])
flagged_analysis
```

```python
#transaction types that are mostly liked to be fraud
flagged_by_type=df[df["isFlaggedFraud"]==1].groupby("type").size().sort_values(ascending=False)
flagged_by_type
```

```python
transfer_analysis=(df[df["type"]=="TRANSFER"].groupby("isFraud")["amount"].agg(["count","mean","median","max"]).round(2))
transfer_analysis
```

```python
df["balance_used_pct"]=((df["amount"]/df["oldbalanceOrg"].replace(0,float("nan")))*100)
df["balance_used_pct"]    
```

```python
transfer_balance_analysis=(df[df["type"]=="TRANSFER"].groupby("isFraud")["balance_used_pct"].agg(["count","mean","median","max"]).round(2))
transfer_balance_analysis
```

```python
fraud_balance_usage=(df[(df["type"]=="TRANSFER")&(df["isFraud"]==1)])
print("Fraudulent_transfers:",len(fraud_balance_usage))
print("used 100% or more:",(fraud_balance_usage["balance_used_pct"]>=100).sum())
```

```python
fraud_balance_pct=((fraud_balance_usage["balance_used_pct"]>=100).sum()/len(fraud_balance_usage)*100)
print("Percentage:",round(fraud_balance_pct,2),"%")
```

```python
zero_balance_analysis=(df.groupby(df["newbalanceOrig"]==0)["isFraud"].agg(["count","mean","sum"]))
zero_balance_analysis
```

```python
zero_balance_analysis["fraud_rate"]=(zero_balance_analysis["mean"]*100).round(2)
zero_balance_analysis
```

```python
repeated_fraud_senders=(df[df["isFraud"]==1].groupby("nameOrig").size().sort_values(ascending=False))
repeated_fraud_senders[repeated_fraud_senders > 1].head(10)
```

```python
fraud_zero_balance=pd.crosstab(df["isFraud"],df["newbalanceOrig"]==0)
fraud_zero_balance
```

```python
zero_balance_by_type=pd.crosstab([df["type"],df["isFraud"]],df["newbalanceOrig"]==0)
zero_balance_by_type
```

```python
fraud_amount_analysis=(df[df["isFraud"]==1].groupby("type")["amount"].agg(["count","sum","mean","median","max"]).sort_values("sum",ascending=False).round(2))
fraud_amount_analysis

                       
```

```python
fraud_sender_analysis=(df[df["isFraud"]==1].groupby("nameOrig")["amount"].agg(["count","sum","mean"]).sort_values("sum",ascending=False).head(10).round(2))
fraud_sender_analysis
```

```python
#compare fraud vs normal transaction
distribution_amount=df.groupby("isFraud")["amount"].quantile([0.25,0.50,0.75,0.90,0.99]).unstack().round(2)
distribution_amount
```

```python
balance_pattern=(df.groupby(["type","isFraud"]).agg(total_transactions=("isFraud","count"),
                                                   zero_balance_transactions=("newbalanceOrig",lambda x:(x==0).sum())).reset_index())
balance_pattern
```

```python
balance_pattern["zero_balance_pct"]=(balance_pattern["zero_balance_transactions"]/balance_pattern["total_transactions"]*100).round(2)
balance_pattern
```

```python
fraud_summary=df.groupby("isFraud").agg(total_transactions=("isFraud","count"),
                                        total_amount=("amount","sum"),
                                        average_amount=("amount","mean")).round(2)
fraud_summary
```

```python
total_transactions=len(df)
fraud_transactions=df["isFraud"].sum()
normal_transactions=total_transactions-fraud_transactions
fraud_rate=fraud_transactions/total_transactions*100
fraud_amount=df.loc[df["isFraud"]==1,"amount"].sum()

print("total_transactions:",total_transactions)
print("fraud_transactions:",fraud_transactions)
print("normal_transactions:",normal_transactions)
print("fraud_rate:",round(fraud_rate,4),"%")
print("total_amount_involved_in_fraud:",round(fraud_amount,2))
```
